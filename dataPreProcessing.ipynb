```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv('./data.csv')
```

```python
data
```

Cleaning data

```python
data = data.drop('id', axis=1)
data
```

```python
Y = data['diagnosis']
Y
```

```python
X = data.drop('diagnosis', axis=1)
X
```

```python
columnToDrop = X.columns.tolist()[-1]
X = X.drop(columns=columnToDrop, axis=1)
X
```

```python
X.dtypes
```

```python
# no of features
n = len(X.columns.tolist())
n
```

```python
# replacing Y with target variables
# Malignant - cancer - 1
# benign - harmless - 0
Y = Y.replace({"M": "1", "B":"0"})
Y = Y.astype(np.int8)
Y
```

Scaling the data

```python
# columns to scale - radius_mean, texture_mean, perimeter_mean, area_mean, radius_se, texture_se, perimeter_se, area_se, radius_worst, texture_worst, perimeter_worst, area_worst, compactness_worst, concavity_worst
columnsToScale = ['radius_mean', 'texture_mean', 'perimeter_mean', 'area_mean', 'radius_se', 'texture_se', 'perimeter_se', 'area_se', 'radius_worst', 'texture_worst', 'perimeter_worst', 'area_worst', 'compactness_worst', 'concavity_worst']
maxies = X[columnsToScale].max().to_list()
length = len(columnsToScale)
for i in range(length):
    X[columnsToScale[i]] = X[columnsToScale[i]]/maxies[i]

X[columnsToScale]
```

Train Test split

```python
XTrain, XTest, YTrain, YTest = train_test_split(X, Y, test_size=0.2)
XTrain.shape
```

```python
X.shape
```

Exporting processed dataset to csv

```python
XTrain.to_csv('XTrain.csv', index=False)
XTest.to_csv('XTest.csv', index=False)
YTrain.to_csv('YTrain.csv', index=False)
YTest.to_csv('YTest.csv', index=False)
```
